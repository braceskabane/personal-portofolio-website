```python
# Define TypeScript Types and Interfaces (as Python classes for validation)

class PersonalInfo:
    def __init__(self, name: str, title: str, subtitle: str, description: str, 
                 profileImage: str, email: str, phone: str, location: str, 
                 website: str, resume: str, social: Dict[str, str]):
        self.name = name
        self.title = title
        self.subtitle = subtitle
        self.description = description
        self.profileImage = profileImage
        self.email = email
        self.phone = phone
        self.location = location
        self.website = website
        self.resume = resume
        self.social = social

class Experience:
    def __init__(self, id: str, title: str, description: str, company: str, 
                 position: str, duration: str, location: str, type: str, 
                 achievements: List[str], technologies: List[str]):
        self.id = id
        self.title = title
        self.description = description
        self.company = company
        self.position = position
        self.duration = duration
        self.location = location
        self.type = type
        self.achievements = achievements
        self.technologies = technologies

class Project:
    def __init__(self, id: str, title: str, description: str, technologies: List[str], 
                 imageUrl: str, demoUrl: str, githubUrl: str, stats: Dict[str, str], 
                 featured: bool, category: str):
        self.id = id
        self.title = title
        self.description = description
        self.technologies = technologies
        self.imageUrl = imageUrl
        self.demoUrl = demoUrl
        self.githubUrl = githubUrl
        self.stats = stats
        self.featured = featured
        self.category = category

class Skill:
    def __init__(self, name: str, level: int, category: str, yearsOfExperience: int):
        self.name = name
        self.level = level
        self.category = category
        self.yearsOfExperience = yearsOfExperience

print("✅ TypeScript interfaces defined as Python classes")
```

```python
# Generate Updated TypeScript Code for MockPortfolioService Class

def generate_typescript_object(obj, indent=2):
    """Convert Python object to TypeScript object string format"""
    if isinstance(obj, dict):
        lines = ["{"]
        for key, value in obj.items():
            formatted_value = generate_typescript_object(value, indent + 2)
            lines.append(f"{' ' * (indent + 2)}{key}: {formatted_value},")
        lines.append(f"{' ' * indent}}}")
        return "\n".join(lines)
    elif isinstance(obj, list):
        if not obj:
            return "[]"
        lines = ["["]
        for item in obj:
            formatted_item = generate_typescript_object(item, indent + 2)
            lines.append(f"{' ' * (indent + 2)}{formatted_item},")
        lines.append(f"{' ' * indent}]")
        return "\n".join(lines)
    elif isinstance(obj, str):
        return f"'{obj}'"
    elif isinstance(obj, bool):
        return "true" if obj else "false"
    elif isinstance(obj, (int, float)):
        return str(obj)
    else:
        return str(obj)

# Generate TypeScript methods
typescript_methods = {}

# 1. getPersonalInfo method
typescript_methods['getPersonalInfo'] = f"""
  async getPersonalInfo(): Promise<PersonalInfo> {{
    await this.delay(300);
    
    return {generate_typescript_object(personal_info_data, 4)};
  }}
"""

# 2. getExperience method  
typescript_methods['getExperience'] = f"""
  async getExperience(): Promise<Experience[]> {{
    await this.delay(600);
    
    return {generate_typescript_object(experience_data, 4)};
  }}
"""

# 3. getSkills method
typescript_methods['getSkills'] = f"""
  async getSkills(): Promise<Skill[]> {{
    await this.delay(400);
    
    return {generate_typescript_object(skills_data, 4)};
  }}
"""

# 4. getProjects method
typescript_methods['getProjects'] = f"""
  async getProjects(): Promise<Project[]> {{
    await this.delay(800); // Simulate API delay
    
    return {generate_typescript_object(projects_data, 4)};
  }}
"""

print("✅ TypeScript methods generated:")
for method_name in typescript_methods.keys():
    print(f"   - {method_name}()")

print("\n📋 Ready to update MockPortfolioService.ts file!")
```

```python
# Test Data Validation and Final Summary

def validate_data():
    """Validate all generated data"""
    validation_results = {
        "personal_info": True,
        "experience": True, 
        "skills": True,
        "projects": True,
        "errors": []
    }
    
    try:
        # Validate PersonalInfo
        if not personal_info.name or not personal_info.email:
            validation_results["personal_info"] = False
            validation_results["errors"].append("PersonalInfo missing required fields")
            
        # Validate Experience
        if len(experiences) < 1:
            validation_results["experience"] = False
            validation_results["errors"].append("Need at least 1 experience entry")
            
        # Validate Skills
        if len(skills) < 5:
            validation_results["skills"] = False
            validation_results["errors"].append("Need at least 5 skills")
            
        # Validate Projects
        if len(projects) < 2:
            validation_results["projects"] = False
            validation_results["errors"].append("Need at least 2 projects")
            
        # Check for featured projects
        featured_projects = [p for p in projects if p.featured]
        if len(featured_projects) < 1:
            validation_results["projects"] = False
            validation_results["errors"].append("Need at least 1 featured project")
            
    except Exception as e:
        validation_results["errors"].append(f"Validation error: {str(e)}")
        
    return validation_results

# Run validation
validation = validate_data()

print("🔍 Data Validation Results:")
print("=" * 50)

if all([validation["personal_info"], validation["experience"], 
        validation["skills"], validation["projects"]]):
    print("✅ ALL VALIDATIONS PASSED!")
else:
    print("❌ Some validations failed:")
    for error in validation["errors"]:
        print(f"   - {error}")

print(f"\n📊 Data Summary:")
print(f"   Personal Info: {'✅' if validation['personal_info'] else '❌'}")
print(f"   Experience: {len(experiences)} entries {'✅' if validation['experience'] else '❌'}")
print(f"   Skills: {len(skills)} skills {'✅' if validation['skills'] else '❌'}")
print(f"   Projects: {len(projects)} projects {'✅' if validation['projects'] else '❌'}")

print(f"\n🎯 Next Steps:")
print("   1. Copy the generated TypeScript methods")
print("   2. Replace corresponding methods in mockPortfolioService.ts")
print("   3. Test the updated portfolio website")
print("   4. Adjust any styling or content as needed")

print(f"\n📁 Files to Update:")
print("   - src/services/portfolio/mockPortfolioService.ts (PRIMARY)")
print("   - src/utils/portfolioChatUtils.ts (SECONDARY)")
print("   - src/app/test-services/page.tsx (OPTIONAL)")
```
